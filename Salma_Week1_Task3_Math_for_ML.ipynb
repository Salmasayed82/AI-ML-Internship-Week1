```python
import numpy as np

data = np.array([10, 20, 20, 30, 40, 50, 60])

mean = np.mean(data)
median = np.median(data)
std = np.std(data)

print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std)
```

```python
total_students = 100
ai_students = 30
probability = ai_students / total_students
print("Probability of selecting an AI student:", probability)
```

```python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])
dot_product = np.dot(vector_a, vector_b)
print("Dot Product:", dot_product)
```

```python
x = 5
derivative = 2 * x
print("Derivative at x =", x, "is", derivative)
```
